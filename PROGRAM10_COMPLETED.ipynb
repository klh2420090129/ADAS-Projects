```python
#Write a program to simulate an Adaptive Cruise Control (ACC) system that adjusts the vehicle speed 
#to maintain a safe following distance from a lead vehicle using a PID controller.

desired_distance = 40   # meters
actual_distance = 30    # meters
our_speed = 80          # km/h
lead_speed = 60         # km/h

Kp = 0.5
Ki = 0.1
Kd = 0.2

accumulated_error = 5
rate_of_error = 2

error = desired_distance - actual_distance

P = Kp * error
I = Ki * accumulated_error
D = Kd * rate_of_error

PID_output = P + I + D
our_speed = our_speed - PID_output

print("Distance Error:", error)
print("P:", P)
print("I:", I)
print("D:", D)
print("PID Output:", PID_output)
print("Lead Vehicle Speed:", lead_speed)
print("Our New Speed:", our_speed)
```
